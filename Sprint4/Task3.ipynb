```
using System;
using System.Collections.Generic;

public interface IElectronicTicket
{
    bool ElectronicTicket{get;set;}

    void CheckElectronicTicket()
    {}
}

public interface ITicket
{
    void PrintTicketInfo();
    void GetTicketDetails();
    void ReserveSeat(int value);
    void RowN(int value);
}

public class Ticket : IElectronicTicket, ITicket
{
    protected int TicketId;
    protected int RowNumber;
    protected int SeatNumber;
    protected int Price;
    public bool ElectronicTicket{get;set;}

    public int TicketID
    {
        get { return TicketId; }
        set { TicketId = value; }
    } 

    void ITicket.RowN(int value) 
    {
        RowNumber = value;
    }

    public int PriceSetting
    {
        get { return Price; }
        set { Price = value; } 
    } 

    public bool CheckElectronicTicket
    {
        get { return ElectronicTicket; }
        set { ElectronicTicket = value; }
    }

    public virtual void GetTicketDetails()
    {
        Console.WriteLine($"TicketId:{TicketId}; RowNumber: {RowNumber}; SeatNumber: {SeatNumber}; Price: {Price}; ElectronicTicket: {ElectronicTicket}");
    }

    void ITicket.ReserveSeat(int value)
    {
        SeatNumber = value;
        Console.WriteLine($"Место номер {SeatNumber} успешно зарезервировано");
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine($"Номер билета: {TicketId}; Номер ряда: {RowNumber}; Номер места: {SeatNumber}, Цена билета: {Price}, Билет электронный - {ElectronicTicket}");
    }

    public void TicketRefund(int value)
    {
        Console.WriteLine($"Возврат билетов осуществлён в количестве {value}");
    }

    public void TicketRefund(bool onWebsite)
    {
        if (onWebsite)
        {
            Console.WriteLine("Возврат оформлен по электронному чеку");
        }
        else
        {
            Console.WriteLine("Возврат оформлен по бумажному чеку");
        }
    }

    public void ChangeTicket(int ticketID, int row, int seat, int price, bool electronicTicket)
    {
        TicketId = ticketID;
        RowNumber = row;
        SeatNumber = seat;
        Price = price;
        ElectronicTicket = electronicTicket;
    }
}

public class ConcertTicket : Ticket
{
    protected string ArtistName;

    public ConcertTicket(int ticketID, int row, int seat, string artistName, int price, bool electronicTicket)
    {
        TicketId = ticketID;
        RowNumber = row;
        SeatNumber = seat;
        ArtistName = artistName;
        Price = price;
        ElectronicTicket = electronicTicket;
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine($"Номер билета: {TicketId}; Номер ряда: {RowNumber}; Номер места: {SeatNumber}, Имя артиста: {ArtistName}, Цена билета: {Price}, Электронный билет - {ElectronicTicket}");
    }
}

public class PlayTicket : ConcertTicket
{
    private string PlayTitle{get;set;}

    public PlayTicket(int ticketID, int row, int seat, string artistName, string playTitle, int price, bool electronicTicket) : base(ticketID,row,seat,artistName,price,electronicTicket)
    {
        PlayTitle = playTitle;
    }

    public override void GetTicketDetails()
    {
        Console.WriteLine($"TicketId:{TicketId}; RowNumber: {RowNumber}; SeatNumber: {SeatNumber}; PlayTitle: {PlayTitle}; Price: {Price}; ElectronicTicket: {ElectronicTicket}");
    }
}

public class TicketCollection<T> where T : Ticket
{
    private List<T> tickets = new List<T>();

    public void Add(T ticket)
    {
        tickets.Add(ticket);
    }

    public void Remove(T ticket)
    {
        tickets.Remove(ticket);
    }

    public void DisplayTickets()
    {
        foreach (var ticket in tickets)
        {
            ticket.PrintTicketInfo();
        }
    }
}

public class TicketService
{
    private readonly ITicket _ticket;

    public TicketService(ITicket ticket)
    {
        _ticket = ticket;
    }

    public void UsingService()
    {
        _ticket.PrintTicketInfo();
        _ticket.GetTicketDetails();
    }
}

Ticket myTicket = new Ticket();
ConcertTicket myCT = new ConcertTicket(2,2,2,"Шопен",500,false);
PlayTicket myPT = new PlayTicket(5,5,5,"Чайковский","Лебединое озеро",1000,true);
TicketCollection<ConcertTicket> ConcertCollection = new TicketCollection<ConcertTicket>();


ConcertCollection.Add(myCT);
ConcertCollection.DisplayTickets();

ITicket firstTicket = new ConcertTicket(4,4,4,"Бах",5000,true);
TicketService firstService = new TicketService(firstTicket);
firstService.UsingService();
```
