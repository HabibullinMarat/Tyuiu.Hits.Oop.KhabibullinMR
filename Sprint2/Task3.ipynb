#### Дополнительное задание

```
using System;

public interface IElectronicTicket
{
    bool ElectronicTicket{get;set;}

    void CheckElectronicTicket()
    {}
}

public class Ticket : IElectronicTicket
{
    protected int TicketId;
    protected int RowNumber;
    protected int SeatNumber;
    protected int Price;
    public bool ElectronicTicket{get;set;}

    public int TicketID
    {
        get { return TicketId; }
        set { TicketId = value; }
    } 

    public int RowN 
    {
        get { return RowNumber; }
        set { RowNumber = value; }
    }

    public int PriceSetting
    {
        get { return Price; }
        set { Price = value; } 
    } 

    public bool CheckElectronicTicket
    {
        get { return ElectronicTicket; }
        set { ElectronicTicket = value; }
    }

    public virtual void GetTicketDetails()
    {
        Console.WriteLine($"TicketId:{TicketId}; RowNumber: {RowNumber}; SeatNumber: {SeatNumber}; Price: {Price}; ElectronicTicket: {ElectronicTicket}");
    }

    public void ReserveSeat(int value)
    {
        SeatNumber = value;
        Console.WriteLine($"Место номер {SeatNumber} успешно зарезервировано");
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine($"Номер билета: {TicketId}; Номер ряда: {RowNumber}; Номер места: {SeatNumber}, Цена билета: {Price}, Билет электронный - {ElectronicTicket}");
    }

    public void TicketRefund()
    {
        Console.WriteLine($"Возврат билета осуществлён");
    }

    public void ChangeTicket(int id, int row, int seat, int price, bool electronicTicket)
    {
        TicketId = id;
        RowNumber = row;
        SeatNumber = seat;
        Price = price;
        ElectronicTicket = electronicTicket;
    }
}

public class ConcertTicket : Ticket, IElectronicTicket
{
    protected string ArtistName;

    public string ArtName 
    {
        get { return ArtistName; }
        set { ArtistName = value; }
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine($"Номер билета: {TicketId}; Номер ряда: {RowNumber}; Номер места: {SeatNumber}, Имя артиста: {ArtistName}");
    }
}

public class PlayTicket : ConcertTicket, IElectronicTicket
{
    private string PlayTitle{get;set;}

    public PlayTicket(int ticketID, int row, int seat, string artistName, string playTitle, int price, bool electronicTicket)
    {
        TicketID = ticketID;
        RowNumber = row;
        SeatNumber = seat;
        ArtistName = artistName;
        PlayTitle = playTitle;
        Price = price;
        ElectronicTicket = electronicTicket;
    }

    public override void GetTicketDetails()
    {
        Console.WriteLine($"TicketId:{TicketId}; RowNumber: {RowNumber}; SeatNumber: {SeatNumber}; PlayTitle: {PlayTitle}");
    }
}

Ticket myTicket = new Ticket();
ConcertTicket myCT = new ConcertTicket();
PlayTicket myPT = new PlayTicket(5,5,5,"Чайковский","Лебединое озеро",1000,true);

myTicket.TicketID = 72;
myTicket.RowN = 5;
myTicket.ReserveSeat(25);
myTicket.GetTicketDetails();
myTicket.PrintTicketInfo();
myTicket.ChangeTicket(1,1,1,500,false);
myTicket.GetTicketDetails();

myCT.ArtName = "Виктор Фридман";
myCT.GetTicketDetails();
myCT.PrintTicketInfo();

myPT.GetTicketDetails();
myPT.PrintTicketInfo();
```
