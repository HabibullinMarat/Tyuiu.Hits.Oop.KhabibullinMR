<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Ticket в C#, который будет представлять билеты на
различные мероприятия. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class Ticket
{
    protected int TicketId;
    protected int RowNumber;
    protected int SeatNumber;

    public int TicketID
    {
        get { return TicketId; }
        set { TicketId = value; }
    } 

    public int RowN 
    {
        get { return RowNumber; }
        set { RowNumber = value; }
    }
    public virtual void GetTicketDetails()
    {
        Console.WriteLine($"TicketId:{TicketId}; RowNumber: {RowNumber}; SeatNumber: {SeatNumber}");
    }

    public void ReserveSeat(int value)
    {
        SeatNumber = value;
        Console.WriteLine($"Место номер {SeatNumber} успешно зарезервировано");
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine($"Номер билета: {TicketId}; Номер ряда: {RowNumber}; Номер места: {SeatNumber}");
    }

    public void ChangeTicket(int id, int row, int seat)
    {
        TicketId = id;
        RowNumber = row;
        SeatNumber = seat;
    }
}

public class ConcertTicket : Ticket
{
    protected string ArtistName;

    public string ArtName 
    {
        get { return ArtistName; }
        set { ArtistName = value; }
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine($"Номер билета: {TicketId}; Номер ряда: {RowNumber}; Номер места: {SeatNumber}, Имя артиста: {ArtistName}");
    }
}

public class PlayTicket : Ticket
{
    private string PlayTitle;

    public string Play_Title
    {
        get { return PlayTitle; }
        set { PlayTitle = value; }
    }

    public override void GetTicketDetails()
    {
        Console.WriteLine($"TicketId:{TicketId}; RowNumber: {RowNumber}; SeatNumber: {SeatNumber}; PlayTitle: {PlayTitle}");
    }
}

Ticket myTicket = new Ticket();
ConcertTicket myCT = new ConcertTicket();
PlayTicket myPT = new PlayTicket();

myTicket.TicketID = 72;
myTicket.RowN = 5;
myTicket.ReserveSeat(25);
myTicket.GetTicketDetails();
myTicket.PrintTicketInfo();
myTicket.ChangeTicket(1,1,1);
myTicket.GetTicketDetails();

myCT.ArtName = "Виктор Фридман";
myCT.GetTicketDetails();
myCT.PrintTicketInfo();

myPT.Play_Title = "Лебединое озеро";
myPT.ChangeTicket(5,5,5);
myPT.GetTicketDetails();
myPT.PrintTicketInfo();
```
