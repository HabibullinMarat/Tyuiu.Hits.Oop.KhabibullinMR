```polyglot-notebook
using System;
using System.Collections.Generic;

// ==================== БАЗОВЫЙ КЛАСС ====================
public class Product
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }

    public Product(string name, decimal price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price:C}, Производитель: {Manufacturer}";
    }

    public virtual void Discount(decimal percent)
    {
        if (percent < 0 || percent > 100)
        {
            Console.WriteLine("Ошибка: процент скидки должен быть в диапазоне 0..100.");
            return;
        }
        Price -= Price * (percent / 100m);
        Console.WriteLine($"Применена скидка {percent}% к продукту \"{Name}\". Новая цена: {Price:C}");
    }

    public virtual void Display()
    {
        Console.WriteLine($"[Товар] {GetInfo()}");
    }
}

// ==================== ЭЛЕКТРОНИКА ====================
public class Electronics : Product
{
    public int WarrantyPeriod { get; set; }

    public Electronics(string name, decimal price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer) => WarrantyPeriod = warrantyPeriod;

    public override string GetInfo() =>
        base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес.";

    public override void Display() =>
        Console.WriteLine($"[Электроника] {GetInfo()}");

    public override void Discount(decimal percent)
    {
        if (WarrantyPeriod > 24 && percent > 15)
        {
            Console.WriteLine($"Товар с гарантией {WarrantyPeriod} мес. — скидка ограничена 15%.");
            percent = 15;
        }
        else if (WarrantyPeriod < 6)
        {
            Console.WriteLine($"Короткая гарантия ({WarrantyPeriod} мес.) — скидка увеличена на 5%.");
            percent += 5;
        }
        base.Discount(percent);
    }
}

// ==================== ОДЕЖДА ====================
public class Clothing : Product
{
    public string Size { get; set; }

    public Clothing(string name, decimal price, string manufacturer, string size)
        : base(name, price, manufacturer) => Size = size;

    public override string GetInfo() =>
        base.GetInfo() + $", Размер: {Size}";

    public override void Display() =>
        Console.WriteLine($"[Одежда] {GetInfo()}");
}

// ==================== КНИГИ ====================
public class Books : Product
{
    public string Author { get; set; }

    public Books(string name, decimal price, string manufacturer, string author)
        : base(name, price, manufacturer) => Author = author;

    public override string GetInfo() =>
        base.GetInfo() + $", Автор: {Author}";

    public override void Display() =>
        Console.WriteLine($"[Книга] {GetInfo()}");
}

// ==================== КОД ВЕРХНЕГО УРОВНЯ (вместо Main) ====================
Console.OutputEncoding = System.Text.Encoding.UTF8;

List<Product> products = new List<Product>
{
    new Electronics("Смартфон Galaxy X", 50000m, "Samsung", 24),
    new Electronics("Пылесос Turbo", 15000m, "Bosch", 3),
    new Clothing("Джинсы Slim Fit", 3500m, "Levi's", "M"),
    new Clothing("Футболка Basic", 1200m, "Nike", "L"),
    new Books("C# для профессионалов", 2500m, "Питер", "Джон Скит"),
    new Books("Чистый код", 1800m, "Питер", "Роберт Мартин")
};

Console.WriteLine("=== ИСХОДНЫЙ СПИСОК ===\n");
foreach (Product p in products) p.Display();

Console.WriteLine("\n=== СКИДКА 20% ===\n");
foreach (Product p in products) p.Discount(20);

Console.WriteLine("\n=== ПОСЛЕ СКИДКИ ===\n");
foreach (Product p in products) p.Display();
```
